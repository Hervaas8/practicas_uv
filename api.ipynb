```python
import time
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

# --- Configuración de la API ---
POLIGPT_API_BASE = "https://api.poligpt.upv.es"
POLIGPT_API_KEY = "[CLAVE_ELIMINADA]" # Tu clave de la UPV

# --- Lista de modelos a interrogar ---
MODELOS_A_TESTEAR = [
    "deepseek-r1:32b",
    "llama3.1:8b",
    "llama3.2:3b",
    "phi4-mini:3.8b",
]

print("🕵️ Iniciando interrogatorio a los modelos...\n" + "-"*50)

for nombre_modelo in MODELOS_A_TESTEAR:
    print(f"\n🤖 Preguntando al endpoint: {nombre_modelo}")
    
    try:
        # Inicializamos el modelo
        llm = ChatOpenAI(
            model=nombre_modelo,
            api_key=POLIGPT_API_KEY,
            base_url=POLIGPT_API_BASE,
            temperature=0.3, # Un poco de temperatura para que hablen natural
            max_tokens=150  # Cortamos rápido para no gastar tokens tontamente
        )
        
        # Le hacemos la pregunta directa
        mensaje = [HumanMessage(content="Hola.")]
        
        inicio = time.time()
        respuesta = llm.invoke(mensaje)
        tiempo = round(time.time() - inicio, 2)
        
        print(f"⏱️ Tiempo de respuesta: {tiempo}s")
        print(f"🗣️ Dice ser: {respuesta.content.strip()}")
        
    except Exception as e:
        print(f"❌ Error al contactar con el modelo: {str(e)}")

print("\n🏁 Interrogatorio finalizado.")
```
